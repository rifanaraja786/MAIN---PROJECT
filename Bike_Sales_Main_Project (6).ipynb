# ***Data Pre-processing Workflow***

I preprocessed  this sales dataset by handling missing entries and standardizing category typos across gender and country fields. count the missing values in each column of the dataset

step -2

***Remove Whitespaces & Standardize the text***

1-Remove accidental leading/trailing whitespaces from column names





```python

df.columns = df.columns.str.strip()
df.columns
```

 After running The .str.strip() function targets every single column name and automatically slices off any accidental spaces hiding at the front or back.

Before running: The Customer_Gender column contained typos and inconsistent values like "Feamle", "Fmale", "Mael", which could cause errors in grouping or analysis.



```python

df['Customer_Gender'] = df['Customer_Gender'].str.strip()

df['Customer_Gender'] = df['Customer_Gender'].replace({
    'Feamle':'F','Fmale':'F','Femal':'F','F':'F',
    'Male':'M','Mael':'M'
})

print(df.iloc[12:18]['Customer_Gender'])
```

After running: All gender values are standardized into  "F" and "M".

Before running: The Country column contained typos and variations like "US", "United States", "Germnay", "Austraila", "United Kindom", which made analysis inconsistent.

```python

df['Country'] = df['Country'].replace({
    'US':'United States','United  States':'United States','United States ':'United States',
    'Germnay':'Germany','Deutschland':'Germany',
    'Austraila':'Australia','Aussie':'Australia',
    'United Kindom':'United Kingdom','UK':'United Kingdom','U.K.':'United Kingdom'
})

print(df.iloc[20:25]['Country'])
```

After running: All these variations were standardized into clean names like "United States", "Germany", "Australia", and "United Kingdom", reducing duplicates and ensuring consistent country categories.

Text columns like Country and Customer_Gender contained accidental padding spaces and multiple spelling variations such as " United States", "Feamle", and "Fmale", which caused the dataset to falsely report 17 unique countries and 7 unique genders due to inconsistent entries.

```python

string_cols = df.select_dtypes(include=['object']).columns
for col in string_cols:
    df[col] = df[col].astype(str).str.strip()

print("Unique Countries:", df['Country'].unique())
print("Unique Genders:", df['Customer_Gender'].unique())
```

After running: This process automatically converts all object columns to clean strings, strips off front and back spaces, and applies explicit category mappings, standardizing values so the data consolidates into 6 unique countries and 2 unique genders for flawless aggregation.

```python

df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df = df.dropna(subset=['Date'])

print(df.loc[5, 'Date'])
print(df.shape)
```

After running:
 The column is fully standardized to datetime64[ns]. As a defensive step, any invalid dates would have been converted to NaT and dropped using .dropna() — in this dataset, df.shape stayed at (5263, 10) before and after this step, confirming no invalid dates actually existed and the Date column was already clean.

Step - 4

***Create Derived Columns***

Before running: The dataset only had the full Date column. The separate Day, Month, and Year columns were either missing or contained incomplete values.

```python
df['Day'] = df['Date'].dt.day.astype(int)

df['Month'] = df['Date'].dt.strftime('%B')

df['Year'] = df['Date'].dt.year.astype(int)

print(df.loc[5, 'Date'])
print(df.loc[5, ['Day','Month','Year']])
print(df.isnull().sum())
```







After running: The Day, Month, and Year columns are successfully extracted from the clean Date column. Since the missing dates are fixed first, the values safely convert into whole numbers (integers) without any code errors. This gives us clean time filters for accurate business reporting.

Before running:  
  The dataset had raw ages but the Age_Group column was incomplete and miscategorized. The goal was to rebuild it accurately using a custom Python function.

```python

df['Customer_Age'] = df['Customer_Age'].astype(int)

def categorized_age(age):
    if age < 25:
        return 'Youth (<25)'
    elif 25 <= age <= 34:
        return 'Young Adults (25-34)'
    elif 35 <= age <= 64:
        return 'Adults (35-64)'
    else:
        return 'Seniors (65+)'


df['Age_Group'] = df['Customer_Age'].apply(categorized_age)

print(df['Age_Group'].isnull().sum())
print(df[['Customer_Age', 'Age_Group']].head(10))
```

After running :
The function fully corrected the Age_Group column, removed blanks and errors, and ensured clean, reliable categories for demographic analysis.

Before running:  
The dataset only had raw dates in the Date column, so it wasn’t possible to directly identify weekend purchases.

```python

df["Is_Weekend"] = df["Date"].dt.dayofweek.isin([5, 6]).astype(int)

print(df[['Date','Is_Weekend']].head(10))
```

After running:  
A new Is_Weekend column was created, marking weekend transactions with 1 and weekday transactions with 0, making it easy to analyze weekend buying behavior.

Before running:  
The dataset only had raw values in the Sales_Order # column, such as ON12345 or ST67890, which did not show whether the order was online or store‑based.

```python

def order_type(order_id):

    if str(order_id).startswith('ON'):
        return 'Online'
    else:
        return 'Store'

df['Order_Type'] = df['Sales_Order #'].apply(order_type)

print(df[['Sales_Order #','Order_Type']].head(10))
```

After running:  
A new column Order_Type was created, classifying each order as “Online” if the ID starts with ON and “Store” otherwise, making the origin of orders clear and easy to analyze.

**Age Distribution**

Before (Aim):  
The aim is to analyze the distribution of customer age to check normality, skewness, and kurtosis. This helps in understanding whether the age variable is symmetrically spread or biased toward certain ranges.

```python
 # Prepare age data, calculate skewness & kurtosis, and set up subplot layout

age_data = df["Customer_Age"].dropna()
age_skewness = skew(age_data)
age_kurtosis = kurtosis(age_data)

chart = make_subplots(
    rows=1, cols=2,
    subplot_titles=(
        "Q-Q Plot: Age vs Normal",
        f"Age Distribution<br>Skew={age_skewness:.2f} | Kurt={age_kurtosis:.2f}"
    )
)

# Create Q-Q plot comparing actual age distribution with theoretical normal distribution

(osm, osr), (slope, intercept, r) = stats.probplot(age_data, dist="norm")
chart.add_trace(go.Scatter(x=osm, y=osr, mode="markers", name="Data", marker=dict(color="blue")), row=1, col=1)
chart.add_trace(go.Scatter(x=osm, y=slope * osm + intercept, mode="lines", name="Normal Line", line=dict(color="red")), row=1, col=1)

# Build histogram of ages and overlay a scaled KDE curve for smooth distribution visualization

bin_counts, bin_edges = np.histogram(age_data, bins=20)
bin_width = bin_edges[1] - bin_edges[0]

x_vals = np.linspace(age_data.min(), age_data.max(), 200)
kde = stats.gaussian_kde(age_data)
kde_scaled = kde(x_vals) * len(age_data) * bin_width

chart.add_trace(go.Bar(x=bin_edges[:-1], y=bin_counts, name="Histogram", marker_color="blue", opacity=0.6), row=1, col=2)
chart.add_trace(go.Scatter(x=x_vals, y=kde_scaled, mode="lines", name="KDE", line=dict(color="red")), row=1, col=2)

chart.update_layout(height=600, width=1200, showlegend=False)
chart.show()
```

After (Finding):  
The skewness value (0.01) confirms the age distribution is essentially symmetric, not skewed toward either end. The kurtosis value (-1.22) indicates a platykurtic (flatter-than-normal) distribution — visible in the histogram as a fairly even spread of order counts across all age ranges rather than a sharp central peak. The Q-Q plot's S-curve shape, flattening at both the youngest and oldest ages, further reflects this short-tailed, evenly-spread distribution. Overall, customer age is broadly uniform across the 18–69 range rather than concentrated around a specific age band.
